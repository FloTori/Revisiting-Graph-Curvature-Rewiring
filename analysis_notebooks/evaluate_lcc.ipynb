```python
import pandas as pd

csv_path = "results_data/lcc_implementation_components.csv"
df = pd.read_csv(csv_path)

implementation_mapping = {
    'topping': 'Topping',
    'networkx': 'NetworkX',
    'pytorch_scipy': 'PyTorch (Scipy)'
}
df["implementation"] = df["implementation"].map(implementation_mapping)

summary = (
    df.assign(method=df["implementation"] + " (" + df["connectiontype"] + ")")
    .pivot_table(index="dataset", columns="method", values="component_size", aggfunc="first")
    .sort_index()
)

summary.T
```

```python
def dataframe_to_latex(df, caption=None, label=None):
    table_df = df.copy()
    index_name = table_df.index.name or "method"

    lines = []
    lines.append("\\begin{table}[htbp]")
    lines.append("\\centering")
    if caption:
        lines.append(f"\\caption{{{caption}}}")
    if label:
        lines.append(f"\\label{{{label}}}")

    column_format = "l" + "r" * len(table_df.columns)
    lines.append(f"\\begin{{tabular}}{{{column_format}}}")
    lines.append("\\hline")

    header = [index_name] + [str(col) for col in table_df.columns]
    lines.append(" \\& ".join(header) + " \\\\")
    lines.append("\\hline")

    for idx, row in table_df.iterrows():
        values = [str(idx)] + [str(int(v)) if pd.notna(v) else "" for v in row.values]
        lines.append(" \\& ".join(values) + " \\\\")

    lines.append("\\hline")
    lines.append("\\end{tabular}")
    lines.append("\\end{table}")

    return "\n".join(lines)

latex_table = dataframe_to_latex(
    summary.T,
    caption="Number of nodes in selected component per dataset and implementation",
    label="tab:lcc_component_sizes",
)

print(latex_table)
```
